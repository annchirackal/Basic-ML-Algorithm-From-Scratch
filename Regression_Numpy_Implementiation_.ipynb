```python
import pandas as pd
import numpy as np
```

```python
### create a an input data set
## two variables x1 and x2 and an out put y

input_df = pd.DataFrame({
    "x1":[1,2,3,4,5],
    "x2":[10,20,30,40,50],
    "y" :[20.5,40.5,180.5,321,500]
})
```

```python
## linear regression :
# 1. estabilish the realation between x an y  --> y = X.beta+bet_zero
# 2. Intaialisze the beta and find y_predictes
# 3. calcuate the loss and cost function
# 4. Calculate the gradient
# 5. make correction in beta
# 6. repeat the process untill error = o or , set nu,ber of epochs
```

```python

class linear_regression():
  def __init__(self,X,y,epochs=20,eta=0.0001):
    self.epochs = epochs
    self.eta = eta
    self.X = X.to_numpy()
    self.y = y.to_numpy().reshape(-1, 1)
    ones = np.ones((self.X.shape[0],1))
    self.X = np.hstack([self.X,ones])
    self.beta = np.zeros((self.X.shape[1],1))

  def train(self):
    for epoch in range(self.epochs):
      print(f'===epoch:{epoch}===')
      self.y_pred =self.X@self.beta

      ## loss = (y- y_pred)
      self.cost = (1/self.X.shape[0]) * np.sum((self.y_pred - self.y)**2)

      gradient = (2/self.X.shape[0])*(self.X.T@(self.y_pred-self.y))

      self.beta -= self.eta*gradient
      print(f"total loss {self.cost} ")




```

```python
model = linear_regression(input_df[['x1','x2']],input_df['y'])
model.train()
```
