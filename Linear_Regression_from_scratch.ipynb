```python
import  pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
class linear_regression():

  def __init__(self):
    self.intercept = 0
    self.slope = 0

  def fit(self,X,y,n_iterations=100, lr = .001):
    self.X = X
    self.y =y
    self.beta_not =0
    self.beta_one = 0
    for _ in range(n_iterations):
    ##y = beta_not + beta_1*x
      y_pred = self.beta_not + self.beta_one*(np.array(self.X))
      ## error = actual - forecast
      error = np.array(self.y)-y_pred
      ### update logic
      ###  beta_0 = old_value + learning_rate* gradient
      #### gradient = -2/n(actual - forecast) --- while using MSE

      ### beta_one  = old_vale + learning_rate *grdaient * x
      self.beta_not = self.beta_not -(lr*2*np.mean(error))
      self.beta_one  = self.beta_one -(lr*2*np.mean(error*self.X))
    self.slope = self.beta_one
    self.intercept = self.beta_not

  def predict(self,x):
    y_predict = (self.slope * np.array(x))+self.intercept
    return y_predict






```

```python
## generate some sythethic data
df = pd.DataFrame({'x':np.linspace(0,50,10)})
df['y'] = 2*df['x']+3
## plot y as a function of x
df.plot(x='x',y='y')
plt.show()
```

```python
model = linear_regression()
```

```python
model.fit(df['x'],df['y'],n_iterations=10000)
```
